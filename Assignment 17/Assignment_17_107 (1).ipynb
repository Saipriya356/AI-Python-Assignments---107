```python
import pandas as pd

df = pd.read_csv("social_media.csv")
df
import pandas as pd
import re
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

def clean_social_media(df):

    # --- Handle Missing Values ---
    df['likes'] = df['likes'].fillna(0)
    df['shares'] = df['shares'].fillna(0)

    # --- Cleaning Text ---
    def clean_text(text):
        if pd.isna(text):
            return ""
        text = text.lower()
        text = re.sub(r'<.*?>', '', text)        # remove html
        text = re.sub(r'[^a-z\s]', '', text)     # remove punctuation/symbols
        words = [w for w in text.split() if w not in ENGLISH_STOP_WORDS]
        return " ".join(words)

    df['clean_text'] = df['post_text'].apply(clean_text)

    # --- Timestamp Features ---
    df['timestamp'] = pd.to_datetime(df['timestamp'])
    df['hour'] = df['timestamp'].dt.hour
    df['weekday'] = df['timestamp'].dt.day_name()

    # --- Remove Duplicate Posts ---
    df = df.drop_duplicates(subset='clean_text')

    return df

cleaned_df = clean_social_media(df)
cleaned_df
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, LabelEncoder

print("================ TASK 3: IoT SENSOR DATA PREPROCESSING ==================\n")

# =========================================================
# 1️⃣ LOAD DATA
# =========================================================
df = pd.read_csv(list(uploaded.keys())[0])

# =========================================================
# 2️⃣ BEFORE CLEANING
# =========================================================
print("--- BEFORE CLEANING ---")
print(df)
print("\n")


# =========================================================
# 3️⃣ CLEANING & PREPROCESSING
# =========================================================

cleaned_df = df.copy()

# Convert timestamp to datetime
cleaned_df['timestamp'] = pd.to_datetime(cleaned_df['timestamp'])

# Handle missing values — forward fill
cleaned_df[['temperature', 'humidity']] = cleaned_df[['temperature', 'humidity']].ffill()

# Remove sensor drift using rolling mean (window = 3)
cleaned_df['temperature_smoothed'] = cleaned_df['temperature'].rolling(window=3, min_periods=1).mean()
cleaned_df['humidity_smoothed'] = cleaned_df['humidity'].rolling(window=3, min_periods=1).mean()

# Normalize using Standard Scaler
scaler = StandardScaler()
cleaned_df[['temp_norm', 'hum_norm']] = scaler.fit_transform(
    cleaned_df[['temperature_smoothed', 'humidity_smoothed']]
)

# Encode sensor IDs
encoder = LabelEncoder()
cleaned_df['sensor_id_encoded'] = encoder.fit_transform(cleaned_df['sensor_id'])


# =========================================================
# 4️⃣ AFTER CLEANING
# =========================================================
print("--- AFTER CLEANING ---")
print(cleaned_df)
print("\n")


# =========================================================
# 5️⃣ TEST CASES
# =========================================================

tests_passed = True

# TEST 1 — No missing values in temperature & humidity
if cleaned_df[['temperature', 'humidity']].isna().sum().sum() == 0:
    pass
else:
    tests_passed = False

# TEST 2 — Normalized columns exist
if "temp_norm" in cleaned_df.columns and "hum_norm" in cleaned_df.columns:
    pass
else:
    tests_passed = False

# TEST 3 — Encoded sensor_id column exists
if "sensor_id_encoded" in cleaned_df.columns:
    pass
else:
    tests_passed = False

# TEST 4 — Rolling mean columns exist
if "temperature_smoothed" in cleaned_df.columns and "humidity_smoothed" in cleaned_df.columns:
    pass
else:
    tests_passed = False


# =========================================================
# 6️⃣ FINAL RESULT
# =========================================================

if tests_passed:
    print("Task 3 Passed All Tests ✔")
else:
    print("❌ Some Tests Failed. Fix Required.")
```

```python
import pandas as pd
import numpy as np
import re
from bs4 import BeautifulSoup
from sklearn.feature_extraction.text import TfidfVectorizer

print("================ TASK 4: MOVIE REVIEWS DATA CLEANING ==================\n")

# =========================================================
# 1️⃣ LOAD DATA
# =========================================================
df = pd.read_csv(list(uploaded.keys())[0])

# =========================================================
# 2️⃣ BEFORE CLEANING (RAW DATA)
# =========================================================
print("--- BEFORE CLEANING ---")
print(df)
print("\n")


# =========================================================
# 3️⃣ CLEANING FUNCTIONS
# =========================================================

def clean_text(text):
    if pd.isna(text):
        return ""
    text = BeautifulSoup(text, "html.parser").get_text()      # Remove HTML tags
    text = text.lower()                                       # lowercase
    text = re.sub(r"[^a-z0-9\s]", "", text)                   # remove punctuation/symbols
    text = re.sub(r"\s+", " ", text).strip()                  # remove extra spaces
    return text


# =========================================================
# 4️⃣ APPLY CLEANING
# =========================================================

cleaned_df = df.copy()

# Clean text column
cleaned_df['clean_text'] = cleaned_df['review_text'].apply(clean_text)

# Fill missing ratings using median
median_rating = cleaned_df['rating'].median()
cleaned_df['rating_filled'] = cleaned_df['rating'].fillna(median_rating)

# Normalize rating (0–10 → 0–1 scale)
cleaned_df['rating_normalized'] = cleaned_df['rating_filled'] / 10

# TF-IDF encoding
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(cleaned_df['clean_text'])

# Convert TF-IDF to DataFrame (top 10 features for display)
tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=vectorizer.get_feature_names_out()
)
tfidf_preview = tfidf_df.iloc[:, :10]   # first 10 columns


# =========================================================
# 5️⃣ AFTER CLEANING (FINAL DATASET)
# =========================================================
print("--- AFTER CLEANING ---")
print(cleaned_df[['review_id', 'clean_text', 'rating_normalized']])
print("\n")

print("--- TF-IDF PREVIEW (First 10 features) ---")
print(tfidf_preview)
print("\n")


# =========================================================
# 6️⃣ TEST CASES
# =========================================================

tests_passed = True

# TEST 1 — HTML removed
if "<" not in cleaned_df['clean_text'].iloc[0]:
    pass
else:
    tests_passed = False

# TEST 2 — No missing ratings after filling
if cleaned_df['rating_filled'].isna().sum() == 0:
    pass
else:
    tests_passed = False

# TEST 3 — Normalized ratings are between 0 and 1
if cleaned_df['rating_normalized'].between(0, 1).all():
    pass
else:
    tests_passed = False


# =========================================================
# 7️⃣ FINAL RESULT
# =========================================================

if tests_passed:
    print("Task 4 Passed All Tests ✔")
else:
    print("❌ Some Tests Failed. Fix Required.")
```
